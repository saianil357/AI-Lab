```python
import random

board = [' '] * 9

def print_board():
    for i in range(0, 9, 3):
        print(f"{board[i]}|{board[i+1]}|{board[i+2]}")

def check_win(p):
    w = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    return any(board[a] == board[b] == board[c] == p for a, b, c in w)

while True:
    print_board()
    pos = int(input("Enter pos (0-8): "))
    if board[pos] != ' ':
        continue
    board[pos] = 'X'

    if check_win('X'):
        print_board()
        print("You win!")
        break
    if ' ' not in board:
        print("Draw!")
        break

    empty = [i for i in range(9) if board[i] == ' ']
    board[random.choice(empty)] = 'O'

    if check_win('O'):
        print_board()
        print("Computer wins!")
        break
```
