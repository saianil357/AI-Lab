```python

import heapq


def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)

            new_state[blank], new_state[new_blank] = (
                new_state[new_blank], new_state[blank]
            )

            neighbors.append(tuple(new_state))

    return neighbors


def manhattan_distance(state, goal):
    distance = 0

    for i in range(9):
        tile = state[i]

        if tile != 0:
            current_row = i // 3
            current_col = i % 3

            goal_index = goal.index(tile)
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            distance += abs(current_row - goal_row)
            distance += abs(current_col - goal_col)

    return distance


def astar(start, goal):
    open_list = []
    heapq.heappush(
        open_list, (manhattan_distance(start, goal), 0, start, [start])
    )

    best_g = {start: 0}

    while open_list:
        f, g, state, path = heapq.heappop(open_list)

        if g != best_g[state]:
            continue

        if state == goal:
            print("Goal reached using A* - Manhattan Distance")
            print("Total moves:", g)

            print("Solution path:")
            for board in path:
                for i in range(0, 9, 3):
                    print(board[i:i+3])
                print()

            return

        for neighbor in get_neighbors(state):
            new_g = g + 1

            if neighbor not in best_g or new_g < best_g[neighbor]:
                best_g[neighbor] = new_g
                h = manhattan_distance(neighbor, goal)
                f = new_g + h

                heapq.heappush(
                    open_list, (f, new_g, neighbor, path + [neighbor])
                )

    print("No solution found.")


def valid_state(state):
    return len(state) == 9 and set(state) == set(range(9))


print("Enter initial state (use 0 for blank):")
start = tuple(map(int, input().split()))

print("Enter goal state (use 0 for blank):")
goal = tuple(map(int, input().split()))

if valid_state(start) and valid_state(goal):
    astar(start, goal)
else:
    print("Invalid input! Enter numbers 0 to 8 exactly once.")
```
