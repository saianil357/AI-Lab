```python

def cost(board):
    c = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                c += 1
    return c

board = [0, 0, 0, 0]

while True:
    current = cost(board)
    print("Current state:", board, "Cost:", current)

    if current == 0:
        print("Goal state reached!")
        break

    best = board[:]
    best_cost = current

    for col in range(4):
        for row in range(4):
            if row != board[col]:
                neighbor = board[:]
                neighbor[col] = row
                c = cost(neighbor)

                print("Neighbor:", neighbor, "Cost:", c)

                if c < best_cost:
                    best = neighbor
                    best_cost = c

    if best_cost == current:
        print("Local minimum reached. Stopping.")
        break

    board = best
```
