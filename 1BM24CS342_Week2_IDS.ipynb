```python
def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col
            new_state = list(state)

            new_state[blank], new_state[new_blank] = (
                new_state[new_blank], new_state[blank]
            )

            neighbors.append(tuple(new_state))

    return neighbors


def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# Depth-Limited Search
def dls(state, goal, limit, path, visited):
    if state == goal:
        return path

    if len(path) - 1 == limit:
        return None

    for neighbor in get_neighbors(state):
        if neighbor not in visited:
            visited.add(neighbor)

            result = dls(
                neighbor, goal, limit,
                path + [neighbor], visited
            )

            if result is not None:
                return result

            visited.remove(neighbor)

    return None


# Iterative Deepening Search
def ids(start, goal):
    limit = 0

    while limit <= 31:
        print("Searching at depth limit:", limit)

        path = dls(start, goal, limit, [start], {start})

        if path is not None:
            print("\nGoal found!")
            print("Number of moves:", len(path) - 1)

            step = 0
            for board in path:
                print("Step", step)
                display(board)
                step += 1

            return

        limit += 1

    print("Goal not found within depth 31.")


# Validate the input
def valid_state(state):
    if len(state) != 9:
        return False

    for number in range(9):
        if state.count(number) != 1:
            return False

    return True


# Main program
print("Enter initial state (use 0 for blank):")
start = tuple(map(int, input().split()))

print("Enter goal state (use 0 for blank):")
goal = tuple(map(int, input().split()))

if not valid_state(start) or not valid_state(goal):
    print("Invalid input! Enter numbers 0 to 8 exactly once.")

else:
    ids(start, goal)
```
