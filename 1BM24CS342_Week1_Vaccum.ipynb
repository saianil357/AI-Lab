```python
location = input("Enter starting location (A/B): ").strip().upper()
status_a = input("Enter status for Room A (Clean/Dirty): ").strip().capitalize()
status_b = input("Enter status for Room B (Clean/Dirty): ").strip().capitalize()

env = {'A': status_a, 'B': status_b}

def is_goal_clean():
    return env['A'] == 'Clean' and env['B'] == 'Clean'

while not is_goal_clean():
    print(f"Agent in Room {location}, Status: {env[location]}")

    if env[location] == 'Dirty':
        print(f"Action: Clean Room {location}")
        env[location] = 'Clean'
    else:
        if location == 'A':
            print("Action: Move Right to B")
            location = 'B'
        else:
            print("Action: Move Left to A")
            location = 'A'

print(f"Goal Reached! Final State: {env}")
```
